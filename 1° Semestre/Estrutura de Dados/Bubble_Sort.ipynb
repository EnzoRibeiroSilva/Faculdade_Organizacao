Código Padrão do Bubble Sort

```python
l = [5, 3, 1, 10, 100]

for i in range(len(l)):
    for j in range(len(l)-i-1):
        if l[j] > l[j+1]:
            l[j], l[j+1] = l[j+1], l[j]
print(l)
```

Exercício 2

```python
v = [1,2,3,4,5]
c = 0

for i in range(len(v)):
    for j in range(len(v)-i-1):
        if v[j] > v[j+1]:
            v[j], v[j+1] = v[j+1], v[j]
    c += 1
print(v)
print("Quantidade de comparações:", c)

# Nenhum elemento é alterado de posição, pois o vetor já está ordendo!
```
