Exercício 2

```python
def area_retangulo(base, altura):
  return round(base * altura)

base = float(input("Digite o valor da base: "))
alt = float(input("Digite o valor da altura: "))

area_retangulo(base, alt)
```

```python
def media_tres(n1, n2, n3):
  media = (n1 + n2 + n3) / 3
  return media

num1 = int(input("Digite um número: "))
num2 = int(input("Digite um número: "))
num3 = int(input("Digite um número: "))

media_tres(num1, num2, num3)
```

Exercício 4

```python
def maior_entre_dois(a, b):
  if a > b:
    return a
  elif a == b:
    return "Os dois são iguais!"
  else:
    return b

num1 = int(input("Digite um número: "))
num2 = int(input("Digite um número: "))

maior_entre_dois(num1, num2)
```

Exercício 6

```python
def contar_vogais(texto):
  cont = 0
  for letra in texto:
    if letra in 'aeiou':
      cont +=1
  return 'Número de vogais:', cont

texto = str(input("Digite algo para a verificação! ")).strip().lower()
contar_vogais(texto)
```

Exercício 7

```python
lista = []

def soma_lista(lista):
  soma = 0

  while True:
    num = int(input("Digite um número: "))
    soma += num
    lista.append(num)
    escolha = input("Quer continuar inserindo elementos? (S/N): ").upper()

    if escolha in ['S', 'SIM']:
      continue
    elif escolha in ['N', 'NÃO', 'NAO']:
      break
    else:
      print("Opção inválida!")
  return f'Lista: {lista}. Soma: {soma}.'


soma_lista(lista)
```

Exercício 8

```python
lista = []

def multiplicar_lista(lista):
  multi = 1

  while True:
    num = int(input("Digite um número: "))
    multi *= num
    lista.append(num)
    escolha = input("Quer continuar inserindo elementos? (S/N): ").upper()

    if escolha in ['S', 'SIM']:
      continue
    elif escolha in ['N', 'NÃO', 'NAO']:
      break
    else:
      print("Opção inválida!")
  return f'Lista: {lista}. Multiplicação: {multi}.'

multiplicar_lista(lista)
```

Exercício 9

```python
def eh_multiplo(a, b):
  if a % b == 0 or b % a == 0:
    return True
  else:
    return False

num1 = int(input("Digite um número: "))
num2 = int(input("Digite um número: "))

eh_multiplo(num1, num2)
```
