Exercício 7

```python
l1 = float(input("Digite o primeiro lado:"))
l2 = float(input("Digite o segundo lado:"))
l3 = float(input("Digite o terceiro lado:"))

lista = [l1, l2, l3]

maior = max(lista)
lista.remove(maior)
somar = sum(lista)

if maior < somar:
  print("O triângulo existe!")
  if l1 == l2 == l3:
    print("O triângulo é equilatero")
  elif l1 == l2 != l3 or l1 == l3 != l2 or l2 == l3 != l1:
    print("O triângulo é isóceles!")
  elif l1 != l2 != l3:
    print("O triângulo é escaleno")
  else:
    print("Algo deu errado, tente novamente!")
else:
  print("O triângulo não pode existir!")
```

Exercício 25

```python
# Feito com inteiros para desconsiderar as moedas do pagamento!
compra = int(input("Digite a compra: "))
valor = int(input("Digite o valor: "))
m_100 = 0
m_50 = 0
m_20 = 0
m_10 = 0
m_5 = 0
m_2 = 0
m_1 = 0

if valor < compra:
  print("Divída existente por falta de débito: R$",compra - valor)
elif valor > compra:
  troco = valor-compra
  print("Troco: R$",troco)
  for i in range(100):
    if troco >= 100:
      m_100 +=1
      troco -= 100
    elif troco >= 50:
      m_50 +=1
      troco -= 50
    elif troco >= 20:
      m_20 +=1
      troco -= 20
    elif troco >= 10:
      m_10 +=1
      troco -= 10
    elif troco >= 5:
      m_5 +=1
      troco -= 5
    elif troco >= 2:
      m_2 +=1
      troco -= 2
    elif troco >= 1:
      m_1 +=1
      troco -= 1
    if troco == 0:
      break
  print("Seu troco possui as seguintes notas:")
  print("Quantidade de Notas de 100:", m_100)
  print("Quantidade de Notas de 50:", m_50)
  print("Quantidade de Notas de 20:", m_20)
  print("Quantidade de Notas de 10:", m_10)
  print("Quantidade de Notas de 5:", m_5)
  print("Quantidade de Notas de 2:", m_2)
  print("Quantidade de Notas de 1:", m_1)
else:
  print("Valor pago corretamente!")
```

Exercício 28

```python
n = int(input("Digite um número para o teste: "))
multi = 0

if n == 0:
  print("0 não é um número perfeito")
else:
  for i in range (1, n):
    if n % i == 0:
      multi += i

  if multi == n:
    print("Esse número é perfeito!")
  else:
    print("Esse número não é perfeito!")
```

Exercício 29

```python
palavra = input("Digite um palavra: ").strip().upper()
lista = []
lista2 = []

for i in range(len(palavra)):
  lista.append(palavra[i])

for letra in range(len(palavra) - 1, -1, -1):
  lista2.append(palavra[letra])

if lista == lista2:
  print("É um palíndromo!")
else:
  print("Não é um palíndromo!")
```
