```python
# Feito com inteiros para desconsiderar as moedas do pagamento!
compra = int(input("Digite a compra: "))
valor = int(input("Digite o valor: "))
m_100 = 0
m_50 = 0
m_20 = 0
m_10 = 0
m_5 = 0
m_2 = 0
m_1 = 0

if valor < compra:
  print("Divída existente por falta de débito: R$",compra - valor)
elif valor > compra:
  troco = valor-compra
  print("Troco: R$",troco)
  for i in range(100):
    if troco >= 100:
      m_100 +=1
      troco -= 100
    elif troco >= 50:
      m_50 +=1
      troco -= 50
    elif troco >= 20:
      m_20 +=1
      troco -= 20
    elif troco >= 10:
      m_10 +=1
      troco -= 10
    elif troco >= 5:
      m_5 +=1
      troco -= 5
    elif troco >= 2:
      m_2 +=1
      troco -= 2
    elif troco >= 1:
      m_1 +=1
      troco -= 1
    if troco == 0:
      break
  print("Seu troco possui as seguintes notas:")
  print("Quantidade de Notas de 100:", m_100)
  print("Quantidade de Notas de 50:", m_50)
  print("Quantidade de Notas de 20:", m_20)
  print("Quantidade de Notas de 10:", m_10)
  print("Quantidade de Notas de 5:", m_5)
  print("Quantidade de Notas de 2:", m_2)
  print("Quantidade de Notas de 1:", m_1)
else:
  print("Valor pago corretamente!")
```

```python
def selection_sort(lista):
    for i in range(len(lista)):
        menor_indice = i
        for j in range(i + 1, len(lista)):
            if lista[j] < lista[menor_indice]:
                menor_indice = j
        lista[i], lista[menor_indice] = lista[menor_indice], lista[i]
    return lista

lista = [64, 25, 12, 22, 11]
print(selection_sort(lista))
```

```python
def soma_de_vetores(vetor1, vetor2):
  if len(vetor1) == len(vetor2):
    for i in range(len(vetor1)):
      for j in range(len(vetor2)):
        if i == j:
          vetor1[i] = vetor1[i] + vetor2[j]
  return vetor1

vetor1 = [1,3,5]
vetor2 = [2,4,6]
soma_de_vetores(vetor1, vetor2)
```
