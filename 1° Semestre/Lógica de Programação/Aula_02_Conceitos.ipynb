### O que é um algoritmo
Um algoritmo é uma sequência finita de passos, organizada para transformar uma entrada em uma saída, seguindo regras claras.
Em computação, a utilidade do algoritmo é permitir que o computador execute o procedimento do mesmo modo, toda vez, sem improviso.
- Determinismo
- Finitude
- Clareza de estados
- Transformação de entrada em saída

### O que é um Sistema
É um conjunto de componentes interconectados que trabalham juntos para alcançar um objetivo específico.

## Variáveis e Tipos de Dados

### O que é uma variável

É um nome associado a um valor guardado na memória.  
Ela representa um estado do programa, isto é, algo que pode mudar ao longo da execução.

#### Analogia
Pense em uma etiqueta colada em uma caixa. A etiqueta é o nome; o conteúdo da caixa é o valor.  
Você pode trocar o conteúdo, mantendo a etiqueta.

### O que é Tipo de Dados
É uma classificação formal que define:
- O conjunto de valores que uma variável pode assumir.
- As operações permitidas sobre esses valores.
- A forma como esses valores são representados na memória.

## 5. Conjuntos

Python possui set e frozenset.

Em Python, `set` e `frozenset` são estruturas de dados do tipo conjunto, baseadas na teoria dos conjuntos da matemática.

Ambos armazenam elementos únicos, não mantêm ordem e permitem operações como união, interseção e diferença.

A diferença central é **mutabilidade**.

#### 1. `set`
`set` é um conjunto mutável.

Isso significa que você pode adicionar e remover elementos após sua criação.

- Características
- Não permite elementos duplicados
- Não possui ordem garantida
- É mutável
- Implementado via tabela hash

##### Exemplo

```python
lista = [1,2,3,4,4]
sem_duplicados = set(lista)
print(sem_duplicados)
```

```python
a = {1,2,3}
b = {3,4,5}

print(a | b) # União
print(a & b) # Interseção
print(a - b) # Diferença
```

#### 2. `frozenset`

`frozenset` é um conjunto imutável.

Depois de criado, não pode ser alterado.

- Características
- Não permite duplicatas
- Não possui ordem garantida
- É imutável
- Pode ser usado como chave de dicionário

##### Exemplo

```python
dicionario = {
    frozenset([1,2]): "Valor"
}
print(dicionario)
```

## Atribuição e atualização

A atribuição usa `=`. Em Python, `=` não significa igualdade matemática, significa "guardar".

Padrão comum de atualização:
- `x = x + 1` para incrementar
- `x += 1` como forma abreviada


```python
contador = 0
contador = contador + 1
contador += 1

print(contador)
```

## Simulação lógica com estados

Programar é, muitas vezes, simular a evolução de estados sob regras.

Exemplo: um placar que muda ao longo do tempo.

```python
c = 0
c -=2

c += 3
print(c)
```

## Operadores aritméticos

Principais operadores:
- `+` soma
- `-` subtração
- `*` multiplicação
- `/` divisão (sempre float)
- `//` divisão inteira
- `%` resto
- `**` potência

Esses operadores constroem expressões, que são avaliadas para produzir um valor.

```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)

print(round(a/b, 2))
```

## Operadores relacionais

Comparam valores e devolvem `bool`:
- `==`, `!=`, `<`, `<=`, `>`, `>=`


```python
x = 7
print(x > 7, x == 7, x !=0, x <7)
```

## Operadores lógicos

Combinam booleanos:
- `and` exige que tudo seja verdadeiro
- `or` exige que pelo menos um seja verdadeiro
- `not` inverte

### Analogia didática
`and` é como "preciso de dois documentos".  
`or` é como "aceito RG ou CNH".  
`not` é como "não pode estar vencido".

```python
idade = 19
td = True
pe = (idade > 18) and td
print('Pode entrar?', pe)
```

#### Avaliação de expressões e precedência
Python segue uma ordem de precedência. Use parênteses para deixar explícito.
Dica: sempre que houver dúvida, use parênteses para documentar intenção.

```python
resultado1 = (2 + 3) * 4
print(resultado1)
```

## Verdadeiro e falso como base da decisão

Decisão computacional se baseia em expressões que avaliam para `True` ou `False`.  
Isso será usado diretamente em `if`, `elif`, `while`.
