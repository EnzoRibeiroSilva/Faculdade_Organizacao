## 1. O que é um dicionário?

Um **dicionário** é uma estrutura de dados que armazena informações no formato:

**chave : valor**

Diferente de listas e tuplas, que organizam dados por posição, o dicionário organiza dados por **associação**.

### Ideia central

Em vez de pensar em posições, no dicionário pensamos em rótulos, como:

- `"nome"`
- `"idade"`
- `"curso"`

Isso torna o dicionário muito útil para representar registros, cadastros, configurações, resultados e dados rotulados.

### Características principais

- é uma estrutura **mutável**;
- armazena pares `chave: valor`;
- o acesso é feito pela **chave**;
- não pode ter **duas chaves iguais**;
- pode armazenar diferentes tipos de dados;
- pode conter listas, tuplas e até outros dicionários.


## 2. Estrutura de um dicionário

A sintaxe básica de um dicionário usa **chaves `{}`**.

Exemplo geral:

```python
{
    "chave1": valor1,
    "chave2": valor2,
    "chave3": valor3
}
```

Cada item do dicionário possui:

- uma **chave**, que identifica o dado;
- um **valor**, que é a informação associada à chave.


````python
aluno = {
    "nome": "Ana",
    "idade": 21,
    "curso": "Ciência de Dados"
}

print(aluno)
print(type(aluno))
````

## 3. Como criar um dicionário

Há várias formas de criar dicionários em Python.


### 3.1 Criação direta com chaves e valores


````python
pessoa = {
    "nome": "Carlos",
    "idade": 30,
    "cidade": "Campinas"
}

print(pessoa)
````

### 3.2 Dicionário vazio

Podemos criar um dicionário vazio e preenchê-lo depois.


````python
dados = {}
print(dados)
print(type(dados))
````

### 3.3 Usando a função `dict()`

Também é possível criar um dicionário com a função `dict()`.


````python
produto = dict(nome="Notebook", preco=3500.00, estoque=12)
print(produto)
````

## 4. Como acessar os dados de um dicionário

O acesso aos dados é feito por meio da **chave**.


````python
aluno = {
    "nome": "Marina",
    "idade": 25,
    "curso": "Engenharia"
}

print(aluno["nome"])
print(aluno["idade"])
print(aluno["curso"])
````

### Pergunta-chave 3

**Como pegar um valor com mais segurança, sem gerar erro caso a chave não exista?**

Resposta: usar `get()`.


````python
print(aluno.get("curso"))
print(aluno.get("telefone"))
print(aluno.get("telefone", "Não informado"))
````

## 5. Não pode ter duas chaves iguais

Em um dicionário, **as chaves devem ser únicas**.

Se uma chave for repetida, o valor mais recente sobrescreve o anterior.


````python
dados = {
    "nome": "Ana",
    "nome": "Beatriz",
    "idade": 20
}

print(dados)
````

Observe que a chave `"nome"` apareceu duas vezes, mas o dicionário manteve apenas o último valor associado a ela.

Portanto:

- pode haver valores repetidos;
- **não pode haver duas chaves iguais** de forma útil, pois a última sobrescreve a anterior.


## 6. Como obter informações de um dicionário

Há algumas perguntas muito comuns quando trabalhamos com dicionários.  
A seguir, cada pergunta é acompanhada do código correspondente.


**Quais são as chaves do dicionário?**

````python
aluno = {
    "nome": "Ana",
    "idade": 21,
    "curso": "Ciência de Dados"
}

print(aluno.keys())
````

**Quais são os valores do dicionário?**

````python
print(aluno.values())
````

**Quais são os pares chave-valor do dicionário?**

````python
print(aluno.items())
````

**Como verificar se uma chave existe?**

````python
print("nome" in aluno)
print("telefone" in aluno)
````

**Quantos itens existem no dicionário?**

````python
print(len(aluno))
````

## 7. Funções e métodos para localizar nomes das chaves

Os métodos mais importantes para localizar ou explorar as chaves são:

- `keys()`, retorna as chaves;
- `in`, verifica se uma chave existe;
- `items()`, retorna chaves e valores juntos.

Esses recursos são fundamentais para navegação e consulta.


````python
config = {
    "tema": "escuro",
    "idioma": "pt-br",
    "fonte": "grande"
}

print("Chaves:", config.keys())
print("Existe a chave 'tema'?", "tema" in config)
print("Existe a chave 'modo'?", "modo" in config)
````

## 8. Como adicionar itens em um dicionário

Para adicionar um novo item, basta atribuir um valor a uma nova chave.


````python
aluno = {
    "nome": "João",
    "idade": 22
}

aluno["curso"] = "Estatística"
print(aluno)
````

## 9. Como modificar itens em um dicionário

Se a chave já existir, a atribuição altera o valor associado.


````python
aluno = {
    "nome": "João",
    "idade": 22,
    "curso": "Estatística"
}

aluno["idade"] = 23
print(aluno)
````

## 10. Como remover itens de um dicionário

Há várias formas de remover elementos.


### 10.1 Usando `del`


````python
aluno = {
    "nome": "João",
    "idade": 22,
    "curso": "Estatística"
}

del aluno["idade"]
print(aluno)
````

### 10.2 Usando `pop()`

O método `pop()` remove a chave informada e ainda retorna o valor removido.


````python
aluno = {
    "nome": "João",
    "idade": 22,
    "curso": "Estatística"
}

valor_removido = aluno.pop("curso")
print("Valor removido:", valor_removido)
print(aluno)
````

### 10.3 Usando `clear()`

Remove todos os itens do dicionário.


````python
dados = {
    "a": 1,
    "b": 2,
    "c": 3
}

dados.clear()
print(dados)
````

## 11. For em dicionário

É muito comum percorrer dicionários com laços `for`.


### 11.1 Percorrendo apenas as chaves


````python
aluno = {
    "nome": "Ana",
    "idade": 21,
    "curso": "Ciência de Dados"
}

for chave in aluno:
    print(chave)
````

### 11.2 Percorrendo as chaves explicitamente com `keys()`


````python
for chave in aluno.keys():
    print(chave)
````

### 11.3 Percorrendo apenas os valores com `values()`


````python
for valor in aluno.values():
    print(valor)
````

### 11.4 Percorrendo chaves e valores com `items()`


````python
for chave, valor in aluno.items():
    print(f"{chave}: {valor}")
````

## 12. Como obter o preço de um produto armazenado em um dicionário?

Esse é um dos usos centrais de dicionários.


````python
produto = {
    "nome": "Mouse",
    "preco": 80.0,
    "estoque": 50
}

preco = produto["preco"]
print("Preço:", preco)
````

### 12.1 Como obter o estoque de um produto?


````python
estoque = produto["estoque"]
print("Estoque:", estoque)
````

## 13. Colocando listas dentro de um dicionário

Um dicionário pode armazenar listas como valores.  
Isso é extremamente útil quando um mesmo campo precisa guardar vários elementos.


````python
aluno = {
    "nome": "Fernanda",
    "notas": [8.5, 7.0, 9.2],
    "disciplinas": ["Python", "Estatística", "Banco de Dados"]
}

print(aluno)
````

### Acessando listas dentro do dicionário


````python
print(aluno["notas"])
print(aluno["notas"][0])
print(aluno["disciplinas"])
print(aluno["disciplinas"][1])
````

**Como calcular a média das notas armazenadas em uma lista dentro do dicionário?**

````python
media = sum(aluno["notas"]) / len(aluno["notas"])
print("Média:", media)
````

### Adicionando elemento à lista que está dentro do dicionário


````python
aluno["notas"].append(10.0)
print(aluno["notas"])
````

## 15. Operações com dados armazenados em dicionários

Os valores armazenados em um dicionário podem ser usados em cálculos e operações.


### 15.1 Soma de valores


````python
vendas = {
    "janeiro": 1200,
    "fevereiro": 1500,
    "marco": 1100
}

total = vendas["janeiro"] + vendas["fevereiro"] + vendas["marco"]
print("Total vendido:", total)
````

### 15.2 Usando `sum()` com `values()`


````python
total = sum(vendas.values())
print("Total vendido:", total)
````

### 15.3 Média de valores armazenados


````python
media = sum(vendas.values()) / len(vendas)
print("Média mensal:", media)
````

### 15.4 Encontrando maior e menor valor


````python
print("Maior venda:", max(vendas.values()))
print("Menor venda:", min(vendas.values()))
````

### 15.5 Encontrando a chave associada ao maior valor


````python
mes_maior = max(vendas, key=vendas.get)
print("Mês com maior venda:", mes_maior)
print("Valor:", vendas[mes_maior])
````

## 16. Atualizando vários itens ao mesmo tempo

O método `update()` permite inserir ou modificar vários pares chave-valor de uma vez.


````python
aluno = {
    "nome": "Ana",
    "idade": 21
}

aluno.update({
    "idade": 22,
    "curso": "Ciência de Dados",
    "cidade": "Jundiaí"
})

print(aluno)
````

## 17. Dicionários aninhados

Um dicionário também pode conter outro dicionário.


````python
turma = {
    "aluno1": {
        "nome": "Ana",
        "nota": 8.5
    },
    "aluno2": {
        "nome": "Bruno",
        "nota": 7.2
    }
}

print(turma)
````

### Acessando dados em dicionários aninhados


````python
print(turma["aluno1"]["nome"])
print(turma["aluno2"]["nota"])
````

## 18. Comparação rápida entre listas e dicionários

| Estrutura | Forma de acesso | Organização |
|---|---|---|
| Lista | por índice | posição |
| Dicionário | por chave | associação |

### Exemplo comparativo


````python
lista = ["Ana", 21, "Ciência de Dados"]

dicionario = {
    "nome": "Ana",
    "idade": 21,
    "curso": "Ciência de Dados"
}

print("Lista, acesso por índice:", lista[0])
print("Dicionário, acesso por chave:", dicionario["nome"])
````

## 19. Erros comuns ao trabalhar com dicionários

Alguns erros frequentes são:

- tentar acessar uma chave que não existe usando `[]`;
- repetir chaves e não perceber a sobrescrita;
- confundir lista com dicionário;
- usar `values()` quando precisava de `keys()` ou `items()`.

### Exemplo de cuidado com chave inexistente


````python
aluno = {
    "nome": "Ana",
    "idade": 21
}

print(aluno.get("telefone"))
print(aluno.get("telefone", "Não cadastrado"))
````

2. apenas valores;
3. chaves e valores.

````python
aluno = {
    "nome": "Lucas",
    "idade": 20,
    "curso": "Estatística"
}

print("Percorrendo apenas as chaves:")
for chave in aluno:
    print(chave)

print("\nPercorrendo apenas os valores:")
for valor in aluno.values():
    print(valor)

print("\nPercorrendo chaves e valores:")
for chave, valor in aluno.items():
    print(f"{chave}: {valor}")
````
