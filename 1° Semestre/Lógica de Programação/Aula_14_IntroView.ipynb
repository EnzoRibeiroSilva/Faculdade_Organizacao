## 1. Gráfico de linhas
##### Sintaxe básica:
``` Python
    plt.plot(x, y)

````python
import matplotlib.pyplot as plt

# Definição dos dados
x = [1,2,3,4,5]
y =[2,4,6,8,10]

plt.plot(x,y) # Criação do gráfico de linha

# Elemento básico de identificação
plt.title('Gráfico de Linha Simples')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')

plt.show() # Exibe o plot
````

````python
import numpy as np

x = np.arange(0,10,0.1) # Cria valores para o eixo das abscissas
y = np.sin(x)

plt.plot(x,y)
plt.title('Gráfico de Seno')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.show()
````

#### Gráfico de linhas (adicionando elementos)

````python
 # Cria uma figura de 12x6 polegadas, dpi cria 80 pontos por polegada
plt.figure(figsize=(12,6), dpi=80) # DPI - Dot Per Inch - É uma medida de densidade que define a qualidade da imagem
plt.plot(x,y, color='purple', label='Seno', linewidth=5)
plt.title('Gráfico da Função Seno', fontsize=17)
plt.xlabel('Eixo X', fontsize=15)
plt.ylabel('Eixo Y', fontsize=15)
plt.legend(loc='upper right')
# plt.grid(True)
plt.show()
````

### 2. Gráficos Multiplos e Subplot
#### Gráficos em uma mesma figura
- Com a função subplots (), você pode desenhar vários gráficos em uma figura:
- Nesse primeiro exemplo, vamos plotar dois sub gráficos de linhas

````python
plt.figure(figsize=(12,6), dpi=80)
plt.subplot(1,2,1)
plt.title('Posição do gráfico 01', fontsize=16)

plt.subplot(1,2,2)
plt.title('Posição do gráfico 01', fontsize=16)

plt.suptitle('Gráficos', fontsize=20)
plt.show()
````

### Gráficos lado a lado

````python
plt.figure(figsize=(12,6), dpi=80)
# Gráfico 1
x = np.arange(0,10,0.1)
y = np.cosh(x)
plt.subplot(1,2,1)
plt.title('Posição do gráfico 01', fontsize=16)
plt.plot(x,y)

# Gráfico 2
x = np.arange(0,10,0.2)
y = np.sin(x)
plt.subplot(1,2,2)
plt.title('Posição do gráfico 02', fontsize=16)
plt.plot(x,y)

plt.suptitle('Gráficos', fontsize=20)
plt.show()
````

### Gráfico um em cima do outro

````python
plt.figure(figsize=(8,8), dpi=80)

x = np.arange(0,10,0.1)
y = np.cosh(x)
plt.subplot(2,1,1)
plt.plot(x,y)

x = np.array([0,1,2,3])
y = np.array([7,6,5,4,])
plt.subplot(2,1,2)
plt.plot(x,y)

plt.suptitle('Gráficos', fontsize=20)
plt.show()
````

### Múltiplos Gráficos com Subplot
- É possível plotar quantos gráficos quiser em uma figura, só precisar informar a quantidade de linhas e colunas que terá a figura, além do índice do gráfico.

````python
plt.figure(figsize=(8,8), dpi=80)

x = np.arange(0,10,0.1)

plt.subplot(2,2,3)
plt.plot(x,np.tan(x))
plt.title('Tangente')

plt.subplot(2,2,1)
plt.plot(x,np.sin(x))
plt.title('Seno')

plt.subplot(2,2,2)
plt.plot(x,np.cos(x))
plt.title('Cosseno')

plt.subplot(2,2,4)
plt.plot(x,np.log(x))
plt.title('Logaritimo')

plt.suptitle('Gráficos', fontsize=20)
plt.show()
````

### 4. Gráficos boxplot com Matplotlib e Seaborn
- Exemplos de boxplot, com a base de dados flores iris, que tem 3 tipos de flores

````python
base_dados = sns.load_dataset('iris')
base_dados.head()
````

````python
plt.figure(figsize=(8,5))
#sns.boxplot(x = 'species', y = 'petal_length', hue='species', data = base_dados, width=0.4, palette=['red', 'blue', 'yellow'])
sns.boxplot(x = 'species', y = 'petal_length', hue='species', data = base_dados, width=0.2, palette='viridis')

plt.title('Boxplot do Comprimento da Pétala por Espécie')
plt.xlabel('Espécie')
plt.ylabel('Comprimento da Pétala')
plt.show()
````

### Mult Boxplot com Matplotlib

````python
setosa = base_dados[base_dados['species'] == 'setosa']['petal_length']
versicolor = base_dados[base_dados['species'] == 'versicolor']['petal_length']
virginica = base_dados[base_dados['species'] == 'virginica']['petal_length']
dados = [setosa, versicolor, virginica]
````

````python
plt.figure(figsize=(8,5))
box = plt.boxplot(dados, patch_artist=True, tick_labels=['setosa','versicolor','virginica'])
# box = plt.boxplot(dados, tick_labels=['setosa','versicolor','virginica'])

# Aplicando cores manualmente
for patch, cor in zip(box['boxes'], ['purple','red','orange']):
    patch.set_facecolor(cor)
    
plt.title('Boxplot do Comprimento da Pétala por Espécie')
plt.xlabel('Espécie')
plt.ylabel('Comprimento da Pétala')
plt.show()
````

#### Formatando por partes

````python
plt.figure(figsize=(8,5))
ticks = ['setosa','versicolor','virginica']
box = plt.boxplot(dados, patch_artist=False, tick_labels= ticks)

cores = ['pink', 'lightblue', 'lightgreen'] # Definição das cores
# 1. Contorno das caixas
for i, bplot in enumerate(box['boxes']):
    bplot.set(color=cores[i], linewidth=3)
# 2. Whiskers (2 por caixa)
cores_whiskers = ['pink','pink','lightblue','lightblue','lightgreen','lightgreen']
for i, whisker in enumerate(box['whiskers']):
    whisker.set(color=cores_whiskers[i], linewidth=3)
# 3. Caps (extremidades)
for i, cap in enumerate(box['caps']):
    cap.set(color=cores_whiskers[i], linewidth=3)
# 4. Medianas
for i, median in enumerate(box['medians']):
    median.set(color=cores[i], linewidth=3)
    
plt.title("Boxplot do Comprimento da Pétala por Espécie")
plt.xlabel("Espécie")
plt.ylabel("Comprimento da Pétala")
plt.show()
````

### Mult Boxplot com Matplotlib Cor

````python
plt.figure(figsize=(8,5))
box = plt.boxplot(dados, notch=True, patch_artist=True, vert=True, tick_labels=['setosa','versicolor','virginica'])
# Notch = ativa o boxplot com entalhe (notched boxplot)

cores = ['pink', 'lightblue', 'lightgreen']
for i, bplot in enumerate(box['boxes']):
    bplot.set(color='gray', linewidth=3)
    bplot.set(facecolor = cores[i])

# Linhas de contorno
for whiskers in box['whiskers']:
    whiskers.set(color='gray', linewidth=3)

# Linhas de contorno
for cap in box['caps']:
    cap.set(color='gray', linewidth=3)

# Linhas de contorno
for median in box['medians']:
    median.set(color='gray', linewidth=3)
    
plt.title('Boxplot do Comprimento da Pétala por Espécie')
plt.xlabel('Espécie')
plt.ylabel('Comprimento da Pétala')
plt.show()
````

### Intervalos de modelagem de séries temporais
- A bibliteca tsmoothie serve para fazer a suavização de séries temporais e detecção de outliers de forma vetorizada.

````python
#!pip install tsmoothie
````

````python
from tsmoothie.smoother import *
import numpy as np
````

````python
x = np.linspace(0,2*np.pi,100)
y = np.sin(x) + np.random.random(100) * 0.2 # Geração do sinal: Seno + Ruído uniforme

# Etapa de suavização
smoother = ConvolutionSmoother( # Instancia um suavizador ór convolução 
    window_len=10, # Tamanho da janela de suavização (5 pontos)
    window_type='ones' # Janela uniforme - média móvel simples
)

smoother.smooth(y) # Aplica a suavização ao sinal y

low, up = smoother.get_intervals(
    'sigma_interval', # Tipo de intervalo (Baseado no desvio padrão)
    n_sigma=2 # Intervalo de + ou - 2 desvios padrões (Aprox 95% sob a normalidade)
)

plt.figure(figsize=(11,6))
plt.plot(smoother.smooth_data[0], linewidth=3, color='blue')
plt.plot(y, '.k')

plt.fill_between(range(len(smoother.data[0])), # Eixo x, índice
                 low[0], # Limite inferior
                 up[0], # Limite superior
                 alpha=0.3) # Tranparência

plt.show()
````

### Gráficos de contorno no Espaço 3D

````python
def f(x,y):
    return np.sin(np.sqrt(x**2+y**2))

x = np.linspace(-6,6,30)
y = np.linspace(-6,6,30)
X,Y = np.meshgrid(x,y)
Z = f(X,Y)

fig = plt.figure(figsize=(10,12))
ax = plt.axes(projection='3d')
ax.contour3D(X,Y,Z,500, cmap='magma')
ax.set_title('Gráficos de Contorno no Espaço 3D', fontsize=18)
plt.show()
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# Configuração inicial
plt.style.use('seaborn-v0_8')
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['axes.titlesize'] = 16
plt.rcParams['axes.titleweight'] = 'bold'

# Carregar dados
iris = load_iris()
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)
iris_df['species'] = iris.target_names[iris.target]

# 1. Cabeçalho do Dashboard
plt.figure(figsize=(14, 1))
plt.text(0.5, 0.5, 'Dashboard - Análise do Conjunto de dados Iris', 
         ha='center', va='center', 
         fontsize=24, fontweight='bold', color='#2e86ab')
plt.axis('off')
plt.show()

# 2. Visualização Geral (Subplots)
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Principais Análises', y=1.02, fontsize=18, fontweight='bold')

# Gráfico 1: Histograma
sns.histplot(data=iris_df, x="sepal length (cm)", hue="species", 
             kde=True, ax=axes[0, 0], palette="husl")
axes[0, 0].set_title('Distribuição do Comprimento da Sépala')

# Gráfico 2: Boxplot
sns.boxplot(data=iris_df, x="species", y="petal width (cm)", 
            hue="species", palette="husl", legend=False, ax=axes[0, 1])
axes[0, 1].set_title('Largura da Pétala por Espécie')

# Gráfico 3: Scatter Plot
sns.scatterplot(data=iris_df, x="sepal width (cm)", y="petal length (cm)", 
                hue="species", style="species", s=100, 
                palette="husl", ax=axes[1, 0])
axes[1, 0].set_title('Relação Sépala vs Pétala')

# Gráfico 4: Heatmap de Correlação
corr = iris_df.iloc[:, :4].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", 
            center=0, ax=axes[1, 1], fmt=".2f")
axes[1, 1].set_title('Matriz de Correlação')

plt.tight_layout()
plt.show()

# 3. Gráfico Adicional (Pair Plot)
plt.figure(figsize=(12, 8))
sns.pairplot(iris_df, hue="species", palette="husl", 
             height=2.5, plot_kws={'s': 80})
plt.suptitle('Relações entre Todas as Variáveis', y=1.02)
plt.show()

# 4. Estatísticas Resumidas
print("\n Métricas Estatísticas por Espécie:")
stats = iris_df.groupby('species').agg(['mean', 'std', 'min', 'max']).round(2)
display(stats.style.background_gradient(cmap='Blues', axis=0))
````
