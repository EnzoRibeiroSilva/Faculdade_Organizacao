## Conceito de repetição

Repetição é executar um mesmo conjunto de passos várias vezes, com pequenas variações.  
Isso aparece em contagem, varredura de dados, simulação e automação de tarefas.

### Estrutura `for` em Python
O `for` itera sobre uma sequência, como um intervalo (`range`) ou uma lista.

## range()

`range(início, fim, passo)` gera uma sequência de inteiros até antes do `fim`.

Exemplos:
- `range(5)` gera 0,1,2,3,4
- `range(1, 6)` gera 1,2,3,4,5
- `range(10, 0, -1)` gera 10 até 1


````python
for i in range(5):
  print(i)
````

````python
for i in range(1, 11):
  print(f'7 X {i} = {7* i}')
````

````python
for i in range(10, 0, -1):
  print(i)
````

## Iteração sobre listas e strings

Listas e strings são sequências. Você pode iterar diretamente pelos elementos.

- Em lista, cada elemento pode ser um número, texto, objeto etc.
- Em string, cada elemento é um caractere.

````python
valores = [10, 20, 30]
for v in valores:
  print(v)
````

````python
texto = ['dados', 'Ciencia', 'IA']

for ch in texto:
  print(ch)
````

## Contagem, acumulação e soma

Padrões clássicos:
- contador: começa em 0 e incrementa
- acumulador: começa em 0 e soma valores

````python
nums = [2, 5, 1, 8]
soma = 0

for n in nums:
  soma += n
print(soma)
````

## Lógica de varredura

Varredura é visitar cada elemento e aplicar uma regra local, por exemplo "se for par, contar".

````python
nums = [2, 5, 1, 8, 10]
pares = 0

for n in nums:
  if n % 2 == 0:
    pares += 1
print('Quantidade de pares: ', pares)
````

## Decisão dentro da repetição

Combinar `for` com `if` é a base de filtragem e contadores condicionais.  
Você varre dados e seleciona apenas o que satisfaz uma condição.

````python
nums = [3, 11, 7, 20, 10, 15]
filtrados = []

for n in nums:
  if n > 10:
    filtrados.append(n)
print(filtrados)
````

## Contadores condicionais

Você pode contar eventos que atendem a uma regra.

Exemplo: quantos alunos foram aprovados (nota >= 7).

````python
notas = [6.5, 7, 8.2, 4.9, 9.1]
aprovados = []
cont = 0

for nota in notas:
  if nota >= 7:
    aprovados.append(nota)
    cont += 1
print(f'Aprovados: {cont}, Notas: {aprovados}')
````

## Encontrar máximo

Padrão: manter o melhor candidato até agora.

````python
nums = [3, 11, 7, 20, 10, 15]
maior = nums[0]

for n in nums[1:]:
  if n >= maior:
    maior = n
print('Maior: ', maior)
````

## for versus while

- `for` é ideal quando você sabe quantas iterações fará, ou quando itera sobre uma sequência.
- `while` é ideal quando você repete até uma condição de parada ficar verdadeira.

Estrutura:
```python
while condição:
    bloco
```

## Condição de parada

A condição do `while` deve mudar ao longo do tempo.  
Se nada muda, você cria um laço infinito.

Boa prática: sempre localize no código qual variável controla a parada.

````python
m = 0
while True:
  n = int(input('Digite um número (0 para sair):'))

  if n == 0:
    break
  m += n
print("Valor acumulado: ", m)
````

## Simulação de processo iterativo

Exemplo: reduzir um estoque até zerar.

````python
estoque = 5

while estoque > 0:
  print('Vendendo 1 unidade, estoque antes:', estoque)
  estoque -= 1
print('Estoque zerado!')
````

## break e continue

- `break` interrompe o laço imediatamente.
- `continue` pula para a próxima iteração.

Use para controle fino, mas com parcimônia, pois pode reduzir legibilidade.

````python
for n in range(1, 200):
  if n % 7 == 0:
    print('Primeiro multiplo de 7: ', n)
    break
````

````python
texto = 'Ciência de Dados'
letras = 0
for ch in texto:
  if ch == ' ':
    continue
  letras +=1
print('Quantidade de letras totais sem espaços: ', letras)
````

## Validação de entrada

Um padrão útil é repetir até o usuário digitar um valor válido.

Isso é frequente em sistemas reais, pois entradas podem estar erradas.

````python
while True:
  senha = int(input('Digite números maiores que 0: '))

  if senha > 0:
    break
  else:
    print('Tente novamente!')
print('Ok, você digitou', senha)
````
