Exercício 1

```python
num = int(input('Digite um número:'))

if num > 0:
    print('Esse número é positivo!')

elif num < 0:
    print('Esse número é negativo!')

else:
    print('Esse é o zero!')
```

Exercício 2

```python
idade = int(input('Digite uma idade:'))

if idade <= 12 and idade >= 0:
    print('Criança')
elif idade > 12 and idade <= 17:
    print('Adolescente')
elif idade >= 18:
    print('Adulto')
else:
    print('Idade inválida')
```

Exercício 6

```python
preco = float(input('Digite um preço R$'))

if preco >= 1 and preco <= 100:
    final = preco * 1.05
    print(f'O valor final é de R${final:.2f}. O juros é de: R${preco * 0.05:.2f}')
elif preco > 100 and preco <= 300:
    final = preco * 1.1
    print(f'O valor final é de R${final:.2f}. O juros é de: R${preco * 0.1:.2f}')
elif preco > 300:
    final = preco * 1.15
    print(f'O valor final é de R${final:.2f}. O juros é de: R${preco * 0.15:.2f}')
else:
    print('O preço inválida!')
```

Exercício 7

```python
l1 = float(input('Digite o lado 1: '))
l2 = float(input('Digite o lado 2: '))
l3 = float(input('Digite o lado 3: '))

if l1 + l2 > l3 and l1 + l3 > l2 and l2 + l3 > l1:
    if l1 == l2 == l3:
        print('Esse triângulo é Equilatero!')
    elif l1 == l2 or l2 == l3 or l1 == l3:
        print('Esse triângulo é isóceles!')
    else:
        print('Esse triângulo é escaleno!')
else:
    print('Esse triângulo não existe!')
```

Exercício Extra: Número primo

```python
num = int(input("Digite um número: "))
div = 0

for i in range(1, num + 1):
  if num % i == 0:
    div += 1

if div > 2:
  print('Esse número não é primo')
elif div <=2:
  print("Esse número é primo!")
else:
  print("Opção inválida!")
```
