```python
# Pilha
capacidade = 5
pilha = []

if len(pilha) < capacidade:
    pilha.append(10)
    print("Empilhado", pilha)
else:
    print("Pilha Cheia", pilha)

# Empilhar outro
pilha.append(20)
print("Empilhado", pilha)

# Desempilhar
if len(pilha) > 0:
    removido = pilha.pop()
    print("Desempilhado", pilha)
else:
    print("Pilha vazia")
```

```python
# Enfileirar
fila = []
fila.append(10)
fila.append(20)
fila.append(30)

print("Fila atual:", fila)

# Desenfileirar
if len(fila) > 0:
    removido = fila.pop(0)
    print("Removido:", removido)
else:
    print("A fila está vazia")

print("Fila atual:", fila)
```

```python
fila = []
while True:
    escolha = int(input('Escolha uma das opções a seguir: \n1 - Enfileirar \n2 - Desenfileirar \n3 - Sair:'))
    if escolha == 1:
        tamanho = int(input("Quantos itens você quer inserir na fila?"))
        if tamanho > 0:
            for i in range(tamanho):
                num = int(input("Digite um número:"))
                fila.append(num)
            print("Fila Atual:", fila)
        else:
            print("Valor inválido para a construção de fila, tente novamente!")
    elif escolha == 2:
        if len(fila) > 0:
            removido = fila.pop(0)
            print("Item removido:", removido)
            print("Fila Atual:", fila)
        else:
            print("Fila vazia!")

    elif escolha == 3:
        print("Fila Final:", fila)
        break
    else:
        print("Escolha inválida, tente novamente!")
```

```python
from collections import deque
lista = deque()

while True:
    escolha = int(input("Escolha uma das opções abaixo: \n1 - Adicionar itens no deque \n2 - Remover itens do deque \n3 - Sair"))
    if escolha == 1:
        escolha_secundaria = int(input("Você gostaria de adicionar no: \n1 - Começo\n2 - Final:"))
        
        if escolha_secundaria == 1:
            itens = int(input("Quantos itens serão adicionados no começo do deque? "))
            
            if itens > 0:
                for i in range(itens):
                    adicionar = int(input("Digite o número desejado:"))
                    lista.appendleft(adicionar)
                print("Lista Atualizada:", lista)
            else:
                print("Digite a opção válida!")
                
        elif escolha_secundaria == 2:
             itens = int(input("Quantos itens serão adicionados no começo do deque? "))
            
            if itens > 0:
                for i in range(itens):
                    adicionar = int(input("Digite o número desejado:"))
                    lista.append(adicionar)
                print("Lista Atualizada:", lista)
            else:
                print("Digite a opção válida!")

        else:
        print("Escolha inválida")

    elif escolha == 2:
        escolha_secundaria = int(input("Você gostaria de remover no: \n1 - Começo\n2 - Final:"))
        if escolha_secundaria == 1:
            lista.pop()
            print("Lista Atualizada:", lista)

        elif escolha_secundaria == 2:
            lista.popleft()
            print("Lista Atualizada:", lista)
        else:
            print("Digite a opção válida!")
            
    elif escolha == 3:
        print("Lista final:", lista)
        break
    else:
        print("Opção inválida")
```
