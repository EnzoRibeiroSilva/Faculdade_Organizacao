## **Fase 1: Fundamentos e Conceitos**

- O que são Tabelas Hash?

É uma estrutura de dados que associa chaves a valores. Então, a partir de um chave, ela faz uma busca rápida e eficiente para encontrar um valor desejado!

- Vantagens

Busca com um O(1), independente do tempo da tabela, a busca será feita no mesmo tempo.

- Desvantagens

Risco de colisão, ou seja, da utilização de duas ou mais chaves resultando no mesmo índice.

- Aplicações práticas

Utilizada para guardar senhas escolhidas de forma mais segura, para criar um indíce para um array, lista telefônica, DNS utilizado para enontrar o IP de um site (É uma tabela hash) etc.

Em linguagens como Python (dict), JavaScript (Object/Map), e Java (HashMap).

Principais conceitos:

- Bucket - Uma unidade de armazenamento na tabela hash

- Colisões - Ocorre quando duas chaves diferentes são mapeadas para a mesma posição (índice)

## **Fase 2: Estrutura e Implementação Básica**

- Objetivo: Implementar uma tabela hash simples com arrays

```python
tamanho_tabela = 10
tabela = [[] for _ in range(tamanho_tabela)]
```

```python
def funcao_hash(chave, tamanho):
    '''Gera um índice numérico baseado na chave.'''
    return abs(hash(chave)) % tamanho

def inserir(tabela, chave, valor):
    '''Insere ou atualiza um valor na tabela hash.'''
    indice = funcao_hash(chave, len(tabela))
    bucket = tabela[indice]

    for par in bucket:
        if par[0] == chave:
            par[1] = valor
            return
    bucket.append([chave, valor])

    if tabela[indice]:
        print(f"Colisão: A chave {chave} tentou o índice {indice}, mas o espaço já estava ocupado por {tabela[indice]}")

def buscar(tabela, chave):
    '''Busca um valor pela chave. Retorna None se não encontrar'''
    indice = funcao_hash(chave, len(tabela))
    bucket = tabela[indice]

    for par in bucket:
        if par[0] == chave:
            return par[1]
    return None

def remover(tabela, chave):
    '''Remove um elemento da tabela hash. retorna True se removido e False se não existir'''
    indice = funcao_hash(chave, len(tabela))
    bucket = tabela[indice]

    for i in range(len(bucket)):
        if bucket[i][0] == chave:
            del bucket[i]
            return True
    return False
```

- Demonstrando colisões

```python
# Simulando um "array" da tabela hash com tamanho fixo
tamanho_tabela = 5
tabela_hash = [[] for _ in range(tamanho_tabela)]

def funcao_hash_simples(chave):
    # Soma os valores ASCII dos caracteres e aplica o módulo
    soma_ascii = sum(ord(char) for char in chave)
    return soma_ascii % tamanho_tabela

def inserir(chave, valor):
    indice = funcao_hash_simples(chave)
    
    # Se a posição já tiver elementos, ocorreu uma COLISÃO
    if tabela_hash[indice]:
        print(f"COLISÃO: A chave '{chave}' tentou o índice {indice}, mas o espaço já está ocupado por: {tabela_hash[indice]}")
    
    tabela_hash[indice].append((chave, valor))

# Inserindo chaves que provavelmente colidirão pelo tamanho reduzido da tabela
inserir("bola", "valor1")
inserir("loba", "valor2") # "bola" e "loba" têm boa chance de cair no mesmo índice
inserir("carro", "valor3")
inserir("arco", "valor4") # "carro" e "arco" podem colidir

print("\nEstado Final da Tabela Hash:")
for i, bucket in enumerate(tabela_hash):
    print(f"Índice {i}: {bucket}")

```

## **Fase 3: Estratégias de Colisão**

- *Encadeamento separado* - A chave é transformada em um índice numérico e os pares de chave-valor são armazenados nesse índice como sublistas para lidar com possíveis condições. Dessa forma a estrutura de dados pode crescer sem limites, dependendo do que for usado!

```python
# 1. Definir o tamanho da tabela hash (preferencialmente um número primo)
tamanho_tabela = 2
tabela = [[] for _ in range(tamanho_tabela)]

# 2. Função Hash
def obter_indice(chave, tamanho):
    return hash(chave) % tamanho

# 3. Função para Inserir ou Atualizar
def inserir(tabela, chave, valor):
    indice = obter_indice(chave, len(tabela))
    bucket = tabela[indice]
    
    # Verificar se a chave já existe para atualizar o valor
    for i in range(len(bucket)):
        chave_existente, _ = bucket[i]
        if chave_existente == chave:
            bucket[i] = (chave, valor)
            return
            
    # Se não existir, adiciona um novo par (encadeamento separado)
    bucket.append((chave, valor))

# 4. Função para Buscar
def buscar(tabela, chave):
    indice = obter_indice(chave, len(tabela))
    bucket = tabela[indice]
    
    # Percorrer a lista para encontrar a chave
    for chave_existente, valor in bucket:
        if chave_existente == chave:
            return valor
            
    return None  # Chave não encontrada

# 5. Função para Remover
def remover(tabela, chave):
    indice = obter_indice(chave, len(tabela))
    bucket = tabela[indice]
    
    for i in range(len(bucket)):
        chave_existente, _ = bucket[i]
        if chave_existente == chave:
            del bucket[i]
            return True  # Removido com sucesso
            
    return False  # Chave não encontrada

# --- Exemplo de Uso ---

inserir(tabela, "Ana", 28)
inserir(tabela, "Carlos", 35)
inserir(tabela, "Beatriz", 22)

# Colisão proposital (se o hash gerar o mesmo índice)
inserir(tabela, "Aan", 40) # 'Ana' e 'Aan' podem cair no mesmo bucket

print("Idade da Ana:", buscar(tabela, "Ana"))
print("Idade da Aan:", buscar(tabela, "Aan"))

tabela
```

- *Sondagem Linear* - Se a posição calculada estiver ocupada, a sondagem linear verifica sequencialmente a próxima posição, até encontrar um espaço!

```python
# Inicializa uma tabela hash de tamanho fixo
tamanho = 10
tabela = [None] * tamanho  # Formato: (chave, valor)

def inserir_linear(chave, valor):
    posicao = hash(chave) % tamanho
    inicio = posicao
    
    while tabela[posicao] is not None:
        if tabela[posicao][0] == chave: # Atualiza se a chave já existir
            tabela[posicao] = (chave, valor)
            return
        
        posicao = (posicao + 1) % tamanho # Sondagem linear
        
        if posicao == inicio: # Tabela cheia
            raise Exception("Tabela hash cheia")
            
    tabela[posicao] = (chave, valor)

def buscar_linear(chave):
    posicao = hash(chave) % tamanho
    inicio = posicao
    
    while tabela[posicao] is not None:
        if tabela[posicao][0] == chave:
            return tabela[posicao][1]
        
        posicao = (posicao + 1) % tamanho
        
        if posicao == inicio: # Deu a volta completa
            break
    return None # Chave não encontrada
```

```python
inserir_linear("Ana", 22)
tabela
```

- *Sondagem Quadrática* - 

```python
# Inicializa uma tabela hash de tamanho fixo
tamanho = 10
tabela_quad = [None] * tamanho 

def inserir_quadratica(chave, valor):
    posicao_base = hash(chave) % tamanho
    i = 0
    
    while True:
        posicao = (posicao_base + i**2) % tamanho # Sondagem quadrática
        
        if tabela_quad[posicao] is None:
            tabela_quad[posicao] = (chave, valor)
            return
        elif tabela_quad[posicao][0] == chave: # Atualiza
            tabela_quad[posicao] = (chave, valor)
            return
            
        i += 1
        if i == tamanho: # Tabela cheia ou ciclo infinito
            raise Exception("Tabela hash cheia ou não conseguiu mapear")

def buscar_quadratica(chave):
    posicao_base = hash(chave) % tamanho
    i = 0
    
    while i < tamanho:
        posicao = (posicao_base + i**2) % tamanho
        
        if tabela_quad[posicao] is None:
            break # Não existe
        if tabela_quad[posicao][0] == chave:
            return tabela_quad[posicao][1]
            
        i += 1
    return None
```

```python
inserir_quadratica("Ana", 30)
```

```python
buscar_quadratica("Ana")
```
