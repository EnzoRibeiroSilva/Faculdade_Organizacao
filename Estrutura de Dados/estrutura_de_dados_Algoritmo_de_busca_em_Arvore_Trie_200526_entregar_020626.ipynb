```python
def merge_sort(list_one):
  if len(list_one) <= 1:
    return list_one

  mid = len(list_one) // 2
  left_half = list_one[:mid]
  right_half = list_one[mid:]

  sorted_left = merge_sort(left_half)
  sorted_right = merge_sort(right_half)

  return merge(sorted_left, sorted_right)


def merge(list_left, list_right):
  result = []
  i = j = 0

  while i < len(list_left) and j < len(list_right):
    if list_left[i] < list_right[j]:
      result.append(list_left[i])
      i += 1
    else:
      result.append(list_right[j])
      j += 1

  result.extend(list_left[i:])
  result.extend(list_right[j:])

  return result
```

```python
def start_trie(*words):
  end_string = "_end_"
  root = {}

  for word in words:
    current_dictionary = root
    for letter in word:
        current_dictionary = current_dictionary.setdefault(letter, {})
    current_dictionary[end_string] = end_string

  return root

def print_trie_tree(node, prefix = "", is_last = True):
  end_string = "_end_"
  connector_char = "└── " if is_last else "├── "
  branch_char = "    " if is_last else "│   "

  keys = merge_sort([k for k in node.keys() if k != end_string])

  for i, key in enumerate(keys):
    is_child_last = (i == len(keys) - 1)
    print(prefix + connector_char + key + ("*" if end_string in node[key] and len(node[key]) == 1 else ""))
    print_trie_tree(node[key], prefix + branch_char, is_child_last)

  if end_string in node and len(node) == 1:
    pass
  elif end_string in node and len(node) > 1 and is_last:
    pass

trie_data = start_trie("casa", "casaco", "cao")

print(trie_data)
print_trie_tree(trie_data)
```
