## 1. O que é uma tupla

Uma **tupla** é uma estrutura de dados **ordenada** e **imutável** em Python.  
Ela permite armazenar vários valores em uma única variável.

### Características principais

- mantém a ordem dos elementos;
- aceita tipos diferentes na mesma estrutura;
- permite valores repetidos;
- é imutável, isto é, não pode ser alterada diretamente após a criação.

Em termos práticos, a tupla é útil quando queremos representar um conjunto de dados que **não deve mudar**, como coordenadas, registros fixos, resultados de funções e pares de valores.


````python
tupla = (10,20,30)
print(tupla)
print(type(tupla))
````

## 2. Como criar uma tupla

Uma tupla pode ser criada com parênteses e elementos separados por vírgulas.


````python
t1 = (1,2,3)
t2 = ('a','b','c')
t3 = (1, "Python", 3.14)

print(f'{t1}, {t2}, {t3}')
````

### 2.1 Tupla com um único elemento

Esse caso exige atenção.  
Para criar uma tupla com um único elemento, é necessário usar uma vírgula.


````python
nao_tupla = (5) 
tupla_unitaria = (5,)
print(nao_tupla, type(nao_tupla))
print(tupla_unitaria, type(tupla_unitaria))
````

### 2.2 Criação sem parênteses

Em Python, a vírgula é o elemento essencial da tupla.  
Por isso, também é possível criar uma tupla sem usar parênteses.


````python
coordenadas = 10,25
print(coordenadas, type(coordenadas))
````

### 2.3 Conversão para tupla

Também é possível usar a função `tuple()` para converter outros iteráveis em tupla.


````python
lista = [1,2,3,4,5]
texto = "Enzo"

print(tuple(lista), tuple(texto))
````

## 3. Como acessar os dados de uma tupla

Os elementos de uma tupla são acessados por **índice**.

- o primeiro elemento está no índice `0`;
- o segundo, no índice `1`;
- índices negativos acessam a partir do final.


````python
frutas = ("Maça", "Banana", "Uva", "Pitanga")
print(frutas[0])
print(frutas[1])
print(frutas[2])
print(frutas[3])
````

### 3.1 Fatiamento

Assim como listas, tuplas também aceitam **slices**.


````python
numeros = (0,1,2,3,4,5,6)
print(numeros[1:4])
print(numeros[:3])
print(numeros[3:])
print(numeros[::2])
````

## 4. Imutabilidade

Uma tupla não pode ter seus elementos alterados diretamente.  
Isso significa que operações de substituição, inserção e remoção não são permitidas.


````python
cores = ("Azul", "Verde", "Vermelho")
print(cores)
# cores[0] = "Amarelo"
````

## 5. Percorrendo uma tupla com `for`

Uma das formas mais comuns de trabalhar com tuplas é percorrê-las com laços de repetição.


````python
linguagens = ("Python", "R", "Julia", "SQL")

for i in linguagens:
    print("Linguagem:", i)
````

### 5.1 Percorrendo com índice

Também é possível percorrer pelos índices com `range(len(...))`.


````python
linguagens = ("Python", "R", "Julia", "SQL")

for i in range(len(linguagens)):
    print(i, linguagens[i])
````

## 6. Unpacking

O **unpacking** permite atribuir os elementos da tupla diretamente a várias variáveis.


````python
pessoa = ("Giovanna", 18, "Analista de Dados")
nome, idade, profissao = pessoa

print(nome, idade, profissao)
````

### 6.1 Unpacking em uma única linha

Esse recurso torna o código mais legível quando a estrutura é fixa.


````python
# coordenada = (12, 8)
# x, y = coordenada

x, y = (12, 8)
print('x =', x)
print('y =', y)
````

### 6.2 Unpacking com `for`

Quando temos uma coleção de tuplas, podemos percorrê-la e desempacotar cada item diretamente no `for`.


````python
alunos = (
    ("Ana", 8.5),
    ("Justino", 9.5),
    ("Lopes", 10),
    ("Enzo P", 7)
)

for nome, nota in alunos:
    print(f'{nome} tirou nota {nota}')
````

### 6.3 Unpacking estendido com `*`

O Python também permite capturar parte dos elementos usando `*`.


````python
valores = (10,20,30,40,50)

primeiro, segundo, *restante = valores
print(primeiro, segundo, restante)

inicio, *meio, fim = valores
print(inicio, meio, fim)

*primeiros, penultimo, ultimo = valores
print(primeiros, penultimo, ultimo)

print(*valores)
````

## 7. Exemplo com acesso aos dados da tupla usando `for` e unpacking

Neste exemplo, cada tupla representa um produto no formato:

`(nome, preço, estoque)`


````python
produto = (
    ("Notebook", 3500, 12),
    ("Mouse", 80, 50),
    ("Teclado", 150, 30)
)

for nome, preco, estoque in produto:
    print(f'Produto: {nome}. Preço: {preco}. Estoque: {estoque}')
````

## 8. Funções e recursos do Python que usam tuplas

Tuplas aparecem com frequência em vários recursos da linguagem.

### Casos importantes

- `enumerate()`;
- múltiplo retorno de funções;
- desempacotamento;
- troca de valores;
- itens de dicionários;
- listas de registros.


### 8.1 `enumerate()`

A função `enumerate()` produz pares no formato de tuplas.


````python
nomes = ("Anderson", "Felipe", "Beatriz")

for item in enumerate(nomes):
    print(item)
````

### 8.2 `enumerate()` com unpacking

Aqui temos um exemplo ainda melhor, pois o laço já separa o índice e o valor.


````python
for indice, nome in enumerate(nomes):
    print(f'Indíce: {indice}: {nome}')
````

### 8.3 Retorno múltiplo de funções

Quando uma função retorna vários valores, o Python organiza esse retorno como tupla.


````python
def calcular(a, b):
    soma = a+b
    produto = a*b
    diferenca = a- b
    return soma, produto, diferenca

resultado = calcular(4,5)
print(resultado)
print(type(resultado))
````

### 8.4 Recebendo o retorno com unpacking

Essa é uma prática muito comum.


````python
num1 = int(input('Digite um número:'))
num2 = int(input('Digite um número:'))

soma, produto, diferenca = calcular(num1, num2)
print(soma, produto, diferenca)
````

Receber resultados desempacotados

````python
print(*resultado)
````

### 8.5 Troca de valores

A troca de valores entre variáveis usa uma forma de empacotamento e desempacotamento.


````python
x = 10
y = 20

x, y = y, x
print(x, y)
````

### 8.6 Itens de dicionários

Ao usar `.items()`, cada item retornado é uma tupla no formato `(chave, valor)`.


````python
idades = {
    "Bruna": 22,
    "Guilherme": 20,
    "Gi Coelho": 25,
}

for nome, idade in idades.items():
    print(nome, idade)
````

## 12. Comparação entre listas e tuplas

Embora sejam parecidas em vários aspectos, listas e tuplas têm finalidades diferentes.

| Aspecto | Lista | Tupla |
|---|---|---|
| Ordem | Sim | Sim |
| Mutabilidade | Sim | Não |
| Sintaxe comum | `[]` | `()` |
| Pode ser alterada | Sim | Não diretamente |
| Métodos de modificação | Muitos | Poucos |
| Uso como chave em dicionário | Não | Sim, se todos os itens forem imutáveis |

### Regra prática

- use **lista** quando os dados precisarem mudar;
- use **tupla** quando os dados forem fixos.


### 12.1 Exemplo direto de comparação

Vamos observar a diferença semântica:

- uma **lista** pode representar uma fila de tarefas em constante mudança;
- uma **tupla** pode representar um ponto geográfico fixo.


## 13. Vantagens das tuplas

As principais vantagens são:

- imutabilidade, que evita alterações acidentais;
- clareza semântica para estruturas fixas;
- compatibilidade com desempacotamento;
- uso como chave em dicionários;
- em geral, menor sobrecarga estrutural do que listas equivalentes.


## 14. Desvantagens das tuplas

As principais limitações são:

- não permitem alteração direta;
- não possuem métodos como `append()`, `remove()` e `sort()`;
- são menos adequadas para dados que mudam constantemente;
- podem exigir recriação total da estrutura para pequenas alterações.
