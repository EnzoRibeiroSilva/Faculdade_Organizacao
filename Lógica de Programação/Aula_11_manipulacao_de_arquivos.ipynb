## 1. Manipulação de Arquivos

Em muitas aplicações, os dados não ficam apenas na memória do programa. Eles precisam ser **armazenados** para uso posterior.

Alguns exemplos:

- salvar relatórios;
- registrar logs de execução;
- armazenar resultados de análises;
- ler bases de dados simples em formato texto;
- trabalhar com imagens, áudios, PDFs e outros arquivos binários.

Na lógica da programação, manipular arquivos é importante porque permite criar programas que **persistem dados**, ou seja, que guardam informação mesmo após o encerramento da execução.

## 2. Conceitos iniciais

Em Python, o trabalho com arquivos normalmente envolve cinco ações fundamentais:

1. **Criação** do arquivo.
2. **Abertura** do arquivo.
3. **Leitura** do conteúdo.
4. **Escrita** de novos dados.
5. **Fechamento** do arquivo.

A função principal usada para isso é `open()`.

A estrutura básica é:

```python
arquivo = open("nome_do_arquivo.txt", "modo")
```

O segundo parâmetro indica o **modo de abertura**.


## 3. Modos de abertura mais usados

| Modo | Significado |
|---|---|
| `r` | leitura |
| `w` | escrita, cria o arquivo ou sobrescreve se já existir |
| `a` | adiciona conteúdo ao final do arquivo |
| `b` | modo binário |
| `t` | modo texto, padrão do Python |
| `x` | criação exclusiva, gera erro se o arquivo já existir |

Exemplos:

- `"r"` → leitura de arquivo texto
- `"w"` → escrita de arquivo texto
- `"rb"` → leitura de arquivo binário
- `"wb"` → escrita de arquivo binário


## 4. Preparando o ambiente

Nesta aula, vamos criar alguns arquivos dentro da própria pasta do notebook. Isso facilita os testes e ajuda a visualizar o resultado de cada operação.


````python
from pathlib import Path

pasta = Path('arquivos_aula') # Caminho, não cria a pasta - Caminho relativo!
pasta.mkdir(exist_ok=True) # mkdir = make directory (Cria a pasta) 
# exists_ok = True - Evita erro caso a pasta já exista

print(f'Pasta de trabalho criada em: {pasta.resolve()}') # resolve() transforma o caminho relativo no completo!
````

# Parte 1, Arquivos de texto

Arquivos de texto armazenam conteúdo legível por humanos, como letras, números e símbolos codificados em formatos como UTF-8.

Exemplos comuns:

- `.txt`
- `.csv`
- `.json`
- `.py`

Agora vamos estudar o ciclo completo dos arquivos de texto: **criação, abertura, leitura, escrita e fechamento**.


## 5. Criação de arquivo de texto

Um arquivo de texto pode ser criado com o modo `w` ou `x`.

- `w` cria o arquivo se ele não existir;
- `w` apaga o conteúdo anterior se o arquivo já existir;
- `x` cria somente se o arquivo ainda não existir.

No exemplo abaixo, vamos criar um arquivo de texto usando `w`.


````python
arquivo_texto = pasta / 'exemplo_texto.txt'
arquivo_texto
````

````python
# Abre o arquivo no modo escrita
# Se ele não existir, será criado
# Se ele já existir, seu conteúdo será substituído
arquivo = open(arquivo_texto, 'w', encoding='utf-8')
# encoding='utf-8' - Define padrão e codificação para suportar acentos!
````

````python
arquivo.write('Primeira linha do arquivo!\n')
arquivo.close() # Fecha o arquivo para garantir que os dados sejam salvos corretamente
````

## 6. Abertura e leitura de arquivo de texto

Depois de criar o arquivo, podemos abri-lo no modo `r` para leitura.

Existem várias formas de leitura:

- `read()` lê todo o conteúdo;
- `readline()` lê uma linha por vez;
- `readlines()` lê todas as linhas e devolve uma lista.


````python
arquivo = open(arquivo_texto, 'r', encoding='utf-8')
conteudo = arquivo.read()
arquivo.close()
print(f'Conteúdo do arquivo: {conteudo}')
````

## 7. Escrita em arquivo de texto

Podemos escrever novas informações em um arquivo de texto de diferentes maneiras:

- com `w`, substituindo tudo;
- com `a`, acrescentando ao final.

No próximo exemplo, vamos adicionar novas linhas ao arquivo sem apagar o conteúdo anterior.


````python
# Esse modo adiciona novas informações no final do arquivo
arquivo = open(arquivo_texto, 'a', encoding='utf-8')
arquivo.write('Segunda linha adicionada\n')
arquivo.write('Terceira linha com novas informações\n')
arquivo.write('Neymar não vai para a copa!\n')
arquivo.close()
````

````python
arquivo = open(arquivo_texto, 'r', encoding='utf-8')
conteudo_atualizado = arquivo.read()
arquivo.close()
print('Conteúdo Atualizado:\n ', conteudo_atualizado)
````

## 8. Leitura linha por linha

Ler linha por linha é útil quando o arquivo é grande ou quando precisamos tratar cada registro separadamente.


````python
arquivo = open(arquivo_texto, 'r', encoding='utf-8')

for numero_linha, linha in enumerate(arquivo, start=1): # start controla o contador da linha
    print(f'Linha {numero_linha}: {linha.strip()}') # strip() tira o "enter" do final da linha
arquivo.close()
````

## 9. Fechamento de arquivos de texto

Fechar o arquivo é uma etapa importante, pois garante:

- liberação de recursos do sistema;
- gravação correta dos dados pendentes;
- redução de erros durante novas operações.

Embora seja possível usar `close()`, em Python é mais seguro utilizar a estrutura `with open(...) as ...`, pois ela fecha o arquivo automaticamente.

O with é uma estrutura do Python usada para **gerenciar recursos automaticamente**, garantindo que eles sejam **abertos e fechados corretamente.**

````python
# with abre, usa e fecha automaticamente
with open(arquivo_texto, 'r', encoding='utf-8') as arquivo:
    conteudo_with = arquivo.read() # Lê o arquivo
print(conteudo_with) # Exibe o conteúdo
````

## 10. Exemplo prático com lista de alunos

Vamos criar um pequeno exemplo mais próximo da rotina de análise de dados. Imagine que desejamos registrar nomes de alunos em um arquivo texto.


````python
arquivo_alunos = pasta / 'alunos.txt'
alunos = ['Justino', 'Enzo', 'Gustavo', 'Gi', 'Eduardo']
with open(arquivo_alunos, 'w', encoding='utf-8') as arquivo:
    for i in alunos:
        arquivo.write(i + '\n')
````

````python
with open(arquivo_alunos, 'r', encoding='utf-8') as arquivo:
    for i in arquivo:
        print(i.strip())
````

# Parte 2, Arquivos binários

Arquivos binários armazenam dados em bytes. Eles não são, em geral, legíveis diretamente por humanos.

Exemplos:

- imagens, como `.png` e `.jpg`;
- áudios, como `.mp3`;
- vídeos, como `.mp4`;
- arquivos compactados, como `.zip`;
- arquivos serializados e formatos proprietários.

Ao trabalhar com arquivos binários em Python, usamos os modos com `b`, como `rb` e `wb`.


## 11. Criação de arquivo binário

Para criar um arquivo binário, podemos escrever uma sequência de bytes. No exemplo a seguir, vamos criar um arquivo binário simples com valores numéricos.


````python
arquivo_binario = pasta / 'dados.bin'
# Criação de uma sequência de bytes
dados = bytes([10,20,30,40,50])
with open(arquivo_binario, 'wb') as arquivo:
    arquivo.write(dados)
    print(f'Arquivo binário criado em: {arquivo_binario}')
````

## 12. Abertura e leitura de arquivo binário

Agora vamos abrir o arquivo binário em modo `rb` e ler seu conteúdo.

Perceba que o resultado será exibido em bytes.


````python
with open(arquivo_binario, 'rb') as arquivo:
    conteudo_binario = arquivo.read()
print(f'Conteúdo Binário: {conteudo_binario}')
print(f'Conteúdo Binário: {list(conteudo_binario)}')
````

## 13. Escrita adicional em arquivo binário

Também é possível acrescentar dados ao final de um arquivo binário usando `ab`.


````python
with open(arquivo_binario, 'ab') as arquivo:
    arquivo.write(bytes([60,70,80,90]))
    
````

````python
with open(arquivo_binario, 'rb') as arquivo:
    dados_atualizados = arquivo.read()
print(dados_atualizados)
print(list(dados_atualizados))
````

## 14. Fechamento de arquivos binários

Assim como nos arquivos de texto, os arquivos binários também devem ser fechados corretamente.

A estrutura `with` continua sendo a opção mais recomendada.


## 15. Comparação entre arquivo texto e arquivo binário

| Característica | Arquivo de texto | Arquivo binário |
|---|---|---|
| Conteúdo | caracteres legíveis | bytes |
| Modos comuns | `r`, `w`, `a` | `rb`, `wb`, `ab` |
| Leitura humana direta | sim | geralmente não |
| Exemplos | `.txt`, `.csv`, `.json` | `.jpg`, `.png`, `.pdf`, `.mp3` |

## 17. Tratamento simples de erros

Ao tentar abrir um arquivo que não existe, o Python gera erro. Podemos tratar isso com `try` e `except`.


````python
arquivo_inexistente = pasta / 'nao_existe.txt'

try:
    with open(arquivo_inexistente, 'r', encoding='utf-8') as arquivo:
        print(arquivo.read())
except FileNotFoundError:
    print("Arquivo não foi encontrado!")
````

## 20. Desafio final

Crie um pequeno programa que:

1. peça ao usuário três nomes de produtos;
2. salve esses nomes em um arquivo texto;
3. leia o arquivo novamente;
4. exiba os nomes armazenados.

Depois, tente adaptar a lógica para registrar códigos numéricos em um arquivo binário.


````python
from pathlib import Path
pasta = Path('desafio_final')
pasta.mkdir(exist_ok=True)

arquivo_produtos = pasta / 'produtos.txt'
````

````python
while True:
    lista = []
    vazio = True
    
    print('Programa de Manipulação de Arquivos\n1 - Inserir produtos\n2 - Visualizar Arquivo Completo\n3 - Exibição de produtos\n4 - Sair')
    escolha = int(input('Escolha uma opção:'))

    if escolha == 1:
        try:
            for i in range(3):
                produto = str(input('Digite o nome do produto: '))
                lista.append(produto)
                if vazio == True:
                    with open(arquivo_produtos, 'w', encoding='utf-8') as arquivo:
                            arquivo.write(f'{lista[i]}\n')
                    vazio = False
                else:
                    with open(arquivo_produtos, 'a', encoding='utf-8') as arquivo:
                            arquivo.write(f'{lista[i]}\n')
        except FileNotFoundError:
            print('O arquivo não existe!')
            
    elif escolha == 2:
        print('\nVisualização do Arquivo Completo!')
        try:
            with open(arquivo_produtos, 'r', encoding='utf-8') as arquivo:
                conteudo = arquivo.read()
            print(conteudo)
        except FileNotFoundError:
            print('O arquivo inexistente!')
    elif escolha == 3:
        print('\nVisualização do Arquivo por Linhas!')
        try:
            with open(arquivo_produtos, 'r', encoding='utf-8') as arquivo:
                for num_linhas, linhas in enumerate(arquivo, start=1):
                    print(f'Linha {num_linhas}: {linhas.strip()}')
        except FileNotFoundError:
            print('Arquivo não existe!')
    elif escolha == 4:
        print('Obrigado por utilizar nosso programa! =)')
        break
    else:
        print('Opção Inválida! Tente novamente-!')
````
