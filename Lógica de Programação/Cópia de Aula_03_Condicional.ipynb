## Condicional Simples: if

## Tomada de decisão

Tomar decisão em um programa é escolher executar um bloco de código ou não, com base em uma condição booleana.

### Estrutura `if`
```python
if condição:
    bloco
```

O bloco só executa quando a condição é verdadeira.

## Condições simples

Condições simples costumam usar operadores relacionais, por exemplo:
- `x > 10`
- `senha == "python"`


````python
saldo = 100
debito = float(input('Digite o valor do débito: '))

if debito < saldo:
  print(f'Saldo inicial: {saldo}')
  saldo -= debito
  print(f'Saldo final: {saldo}')
print('Fim')
````

````python
senha = input('Senha: ')
if senha == 'python':
  print('Acesso liberado')
````

## Decisão humana e decisão computacional

Decisão humana pode ser vaga, por exemplo "se estiver frio, leve casaco".  
Decisão computacional precisa de regra mensurável, por exemplo "se temperatura < 18".


## Estrutura `if / else`

Quando existem dois caminhos mutuamente exclusivos, use `else`.

```python
if condição:
    caminho_A
else:
    caminho_B
```

A ideia é garantir que exatamente um caminho seja escolhido.

<div>
<img src=attachment:7f6bc8ae-ea25-4928-add1-2c0645452fa3.png style="height:500px; width:auto;"/>
</div>

````python
saldo = 100
debito = float(input('Digite o valor do débito: '))

if debito < saldo:
  print(f'Saldo inicial: {saldo}')
  saldo -= debito
  print(f'Saldo final: {saldo}')
else:
  diferenca = debito - saldo
  print('Saldo insuficiente! \nDeposite', diferenca, 'reais para continuar')
````

````python
senha = input('Senha: ')
if senha == 'python':
  print('Acesso liberado')
else:
  print('Acesso negado, senha incorreta!')
````

## Casos mutuamente exclusivos e erros comuns

Erros típicos:
- condição invertida
- esquecimento de atualizar variável antes de decidir
- usar `==` quando deveria usar `>=` ou `<=`

Uma boa prática é testar casos de borda, por exemplo 17, 18, 19.

````python
for idade in [17, 18, 19]:
  msg = 'maior' if idade >= 18 else 'Menor'
  print(idade, '->', msg)
````

## Estrutura `if / elif / else`

Quando há mais de dois caminhos, use `elif` para encadear decisões.

```python
if cond1:
    ...
elif cond2:
    ...
else:
    ...
```

A primeira condição verdadeira determina o caminho, as demais são ignoradas.

## Prioridade de condições

A ordem importa. Coloque primeiro as regras mais específicas ou mais restritivas.  
Um bom método é escrever a tabela de decisão antes.

## Tabela de decisão

Crie uma tabela simples para validar lógica:

| Faixa | Condição | Saída |
|---|---|---|
| A | nota >= 9 | "A" |
| B | 7 <= nota < 9 | "B" |
| C | 5 <= nota < 7 | "C" |
| D | nota < 5 | "D" |


````python
nota = float(input('Digite uma nota: '))

if nota >= 9:
  print('A')
elif nota >= 7:
  print('B')
elif nota >= 5:
  print('C')
else:
  print('D')
````
