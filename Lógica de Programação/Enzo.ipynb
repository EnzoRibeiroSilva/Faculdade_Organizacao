# ✅ **Verificações**

**Explicação**: Durante a Análise exploratória de Dados (EDA), é necessário verificar se existem informações nulas ou duplicadas no DataFrame. Essa etapa é essencial porque os dados que forem obtidos com ignorância dessa parte pode afetar o resultado final de uma forma extremamente prejudicial.

Portanto, vamos verificar os espaços Vazios/Valores nulos por colunas a seguir!

```python
df.isnull().sum()
```

- *df.isnull()* - Procura por valores nulos no DataFrame
- *.sum()* - Faz a contagem de todos os valores procurados, para poder mensurar a quantidade!

**Observação**: Como o valor foi 0, isso significa que não existem valores nulos/vazios nesse DataFrame, todas as linhas de todas as colunas foram preenchidas!

```python
df.duplicated().sum()
```

- **df.duplicated().sum()** - Esse comando procura por linhas duplicadas e contabiliza elas, portanto temos 4014 linhas duplicadas aqui.

```python
df[df.duplicated()]
```

- **df[df.duplicated()]** - Traz todas as linhas que são duplicadas no DataFrame, sendo possível analisar todas as informações delas.

```python
df_reduzido = df.drop_duplicates().copy()
df_reduzido.duplicated().sum()
```

**Explicação**:
- *df_reduzido = df.drop_duplicates().copy()* - Aqui é feita uma cópia do DataFrame original, essa cópia exclui todas as linhas duplicadas do Banco.
- *df_reduzido.duplicated().sum()* - Para garantir que a nossa cópia (Feita para não gerar problemas no Banco original) cumpriu a função, verificamos a quantidade de linhas duplicadas. (Como retornou 0, isso significa que não existem mais linhas duplicadas no nosso novo DataFrame)

```python
df.T.duplicated().sum()
```

## **Observações da Análise Inicial**:

- *df.T.duplicated().sum()* - Verificação das colunas duplicadas no DataFrame original

- Não existem **valores nulos** no Banco de Dados
- Antes da limpeza, 4014 linhas eram duplicadas, agora **não existem mais linhas duplicadas**.
- **Duplicação de linhas X Duplicação de colunas** - Quando uma linha é duplicada, temos dados presentes em todas as colunas que são duplicados. A duplicação delas é complicada de lidar, porque ela pode atrapalhar o resultado final da análise, com valores e informações que geram desvios incorretos.

  Quando uma coluna é duplicada, isso significa que ocorre uma **redundância** do Banco, porque é o mesmo dado em um frequência desnecessária e prejudicial. A duplicação de colunas ocorre quando existe algum erro de código ou caso seja as mesmas informações armazenadas em duas colunas que são também, a mesma coisa.
- **Dados duplicados afetam** médias, contagens e conclusões, cada valor pode gerar uma alteração que pode ser pequena, mas que gera diferenças do real que podem atrapalhar dependendo do tamanho do banco, sendo uma solução de tamanho empresarial por exemplo.

# 🕐 **Intervalo Temporal**

📅 **Módulo Datetime**: É uma biblioteca padrão essencial para manipulação, formatação e cálculos envolvendo datas e horários. Ele transforma dados temporais (frequentemente carregados como texto/string) em objetos manipuláveis, permitindo operações aritméticas como encontrar a diferença entre dias ou somar horas. Na análise de dados, ele é crucial para estruturar séries temporais, permitindo extrair componentes específicos (ano, mês, dia da semana) e realizar análises baseadas em períodos

```python
df_pt_br['ano_trabalho'] = pd.to_datetime(df_pt_br['ano_trabalho'], format='%Y')
print('Início:', df_pt_br['ano_trabalho'].min())
print('Fim:', df_pt_br['ano_trabalho'].max())
```

 **Explicação**:
 - **df_pt_br['ano_trabalho'] = pd.to_datetime(df_pt_br['ano_trabalho'], format='%Y')** - tem como objetivo converter uma coluna de texto (string/object) que contém apenas anos para o tipo de dado de data (datetime64) do Pandas, garantindo que o formato seja interpretado corretamente como um ano de 4 dígitos.
 - **print('Início:', df_pt_br['ano_trabalho'].min())** - Mostrar o menor ano do DataFrame, descobrindo o ínicio dos dados.
 - **print('Fim:', df_pt_br['ano_trabalho'].max())** - Mostrar na tela o maior ano do DataFrame, descobrindo o final dos dados.

# **Tipos de Gráficos**

## **Gráfico de Barras**
Tipo de variável: Categóricas, tanto nominais, quanto ordinais.

Permite Visualizar: Comparações entre categorias/Frequência de categorias.

Quando usar: Os gráficos de barras e colunas devem ser utilizados quando você tem dados classificados como nominais (dados categóricos que apresentam uma ordem natural) e ordinais (dados categóricos que não apresentam uma ordem natural).

### **Exemplo**

```python
import seaborn as sns

traducao = {
    'Senior': 'Sênior',
    'Mid-level': 'Pleno',
    'Entry-level': 'Júnior',
    'Executive': 'Executivo'
}

df_pt_br['nivel_experiencia'] = df_pt_br['nivel_experiencia'].replace(traducao)

frequencia = df_pt_br['nivel_experiencia'].value_counts()
```

```python
ordem_desejada = ['Júnior', 'Pleno', 'Sênior', 'Executivo']

plt.figure(figsize=(8,6))

sns.barplot(data=df_pt_br, order=ordem_desejada)

plt.bar(frequencia.index, frequencia.values)

plt.title('Frequência por nível de experiência')
plt.xlabel('Nível de experiência')
plt.ylabel('Quantidade')
plt.show()

# Anotação
# sns.barplot(
#     x='nivel_experiencia',
#     y='salario_dolar',
#     data=df_pt_br,
#     estimator='mean',
#     errorbar=None
# )
```

**Explicação**

- .value_counts() - Conta a frequência de cada categoria na coluna.
- frequencia - Armazena 'int', pois é onde a frequência realizada pelo .value_counts() é armazenada.
- (frequencia.index, frequencia.values) - Correspondem a categorias e frequências (Categorias e altura do gráfico)
- plt.bar() - Esse comando cria o gráfico
- plt.title(), plt.xlabel(), plt.ylabel() - Coloca o nome do título, eixo x e eixo y.
- plt.show() - Mostra o gráfico na tela


**Ordenar**
- Para isso, armazene as categorias em uma lista, use o comando: sns.barplot(data=df_pt_br, order=ordem_desejada), onde você colocará o DataFrame e a lista na sequência.

### **Discussão**

**Observações**
- Sênior é a categoria com a maior quantidade de dados no DataFrame.
- O gráfico de barras permite visualizar a quantidade de pessoas em cada categoria no banco, portanto, facilita a comparação entre as mesmas.
- O gráfico de barras é mais interessante para explorar variavéis categóricas, porém caso seja possível unir as duas, talvez seja interessante, por enquanto não é.


## **Histograma**

## **Gráfico de Linhas**

Tipo de variável: Variáveis numéricas ao longo de um período de tempo

Permite visualizar: Evolução dos dados, tendências de mercado, variação em um período de tempo.

Quando usar: Ele conecta pontos de dados individuais com segmentos de reta, facilitando a identificação rápida de aumentos, diminuições, picos ou estabilização de métricas.

### **Exemplo**

```python
media_salario = df_pt_br.groupby('ano_trabalho')['salario_dolar'].mean()

plt.figure(figsize=(11, 6))
plt.plot(media_salario.index, media_salario.values, marker='o')

plt.title("Mediana Salarial Por Ano")
plt.xlabel("Ano")
plt.ylabel("Salario em USD")
plt.show()
```

**Explicação**
- .groupby('ano_trabalho') - Agrupa o ano de trabalho com o salário em dolár para o cálculo da mediana.
- mean() é para média, median() é mediana. Podemos utilizar mean em um caso onde os dados são muito bem distribuídos, porém a mediana é melhor no nosso caso, pois existem outliers e os dados são predominantes em uma faixa e poucos em outras.
- .index - Representa os valores presentes no eixo x.
- .values - Representa os valores no eixo y.
- É importante agregar dados por ano para simplificar os dados em um fluxo temporal, identificando melhor picos, quedas ou estabilidades de períodos.

### **Discussão**

- O salário mediano apresenta aumento ao longo dos anos.
- O gráfico é adequado para analisar médias, modas e medianas ao longo de um período de tempo determinado.
- Fatores externos que podem afetar a evolução: Popularidade da área de dados, possibilidade de home office durante a pandemia, procura de empregos relacionados a tecnologia, alta demanda por profissionais de dados nas empresas.

## **Gráfico de Pizza**

Tipo de variável: Variáveis categóricas.

Permite visualizar: Porcentagens de cada categoria.

Quando usar: Com a menor quantidade possível de categorias, para simplificar o resultado e não gerar poluição visual.

```python
traduzir_modelos = {
    'In-person':'Presencial',
    'Hybrid':'Híbrido',
    'Remote':'Remoto'
}

df_pt_br['modelo_trabalho'] = df_pt_br['modelo_trabalho'].replace(traduzir_modelos)
```

```python
frequencia = df_pt_br['modelo_trabalho'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(frequencia.values, labels=frequencia.index, autopct='%1.1f%%', startangle=90, wedgeprops={'edgecolor': 'white', 'linewidth': 1})
plt.title("Proporção Dos Modelos Do Trabalho")
plt.show()
```

**Explicação**
- value_conts - Vai realizar a frequência da informação da coluna selecionada
- labels - Seria o eixo Y, onde temos as categorias.
- autopct - Formatação da porcentagem
- plt.pie() - Cria o gráfico de pizza.

### **Discussão**

- O modelo de trabalho mais frequente é o Presencial.
- O gráfico de pizza facilita a visualização dos dados em forma de porcentagem
- Em situações em que existem muitas categorias o gráfico de pizza não é indicado!
