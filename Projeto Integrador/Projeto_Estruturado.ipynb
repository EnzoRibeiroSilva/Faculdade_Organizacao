# **Modelos de Análise de Sentimento**

**Vader**: (Valence Aware Dictionary and sEntiment Reasoner) é uma ferramenta de Processamento de Linguagem Natural (PLN) baseada em léxico, amplamente utilizada para classificar o sentimento de textos (positivo, negativo ou neutro). Ela é especialmente eficiente para analisar gírias e emojis comuns em mídias sociais.


**TextBlod**: Uma biblioteca Python intuitiva para Processamento de Linguagem Natural (PLN), ideal para tarefas rápidas de análise textual. Construída sobre as bibliotecas NLTK e Pattern, ela destaca-se pela facilidade de uso em recursos como análise de sentimentos e correção ortográfica.


**leIA**: (Léxico para Inferência Adaptada) é um fork do léxico e uma ferramenta para análise de sentimentos.


**Transforms**: Desenvolvida pela Hugging Face, é o padrão da indústria para trabalhar com modelos de Inteligência Artificial de última geração, como GPT, BERT e LLaMA. Ela fornece ferramentas simples para processamento de linguagem natural, visão computacional e áudio.
