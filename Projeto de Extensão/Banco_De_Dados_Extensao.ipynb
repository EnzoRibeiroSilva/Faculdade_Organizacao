# **Verificações**

Explicação: Durante a Análise exploratória de Dados (EDA), é necessário verificar se existem informações nulas ou duplicadas no DataFrame. Essa etapa é essencial porque os dados que forem obtidos com ignorância dessa parte pode afetar o resultado final de uma forma extremamente prejudicial.

Portanto, vamos verificar os espaços Vazios/Valores nulos por colunas a seguir!

```python
df.isnull().sum()
```

- *df.isnull()* - Procura por valores nulos no DataFrame
- *.sum()* - Faz a contagem de todos os valores procurados, para poder mensurar a quantidade!

**Observação**: Como o valor foi 0, isso significa que não existem valores nulos/vazios nesse DataFrame, todas as linhas de todas as colunas foram preenchidas!

```python
df.duplicated().sum()
```

- df.duplicated().sum() - Esse comando procura por linhas duplicadas e contabiliza elas, portanto temos 4014 linhas duplicadas aqui.

```python
df[df.duplicated()]
```

- **df[df.duplicated()]** - Traz todas as linhas que são duplicadas no DataFrame, sendo possível analisar todas as informações delas.

```python
df_reduzido = df.drop_duplicates().copy()
df_reduzido.duplicated().sum()
```

**Explicação**:
- df_reduzido = df.drop_duplicates().copy() - Aqui é feita uma cópia do DataFrame original, essa cópia exclui todas as linhas duplicadas do Banco.
- df_reduzido.duplicated().sum() - Para garantir que a nossa cópia (Feita para não gerar problemas no Banco original) cumpriu a função, verificamos a quantidade de linhas duplicadas. (Como retornou 0, isso significa que não existem mais linhas duplicadas no nosso novo DataFrame)

```python
df.T.duplicated().sum()
```

## **Observações da Análise Inicial**:
- df.T.duplicated().sum() - Verificação das colunas duplicadas no DataFrame original
- Não existem colunas duplicadas no Dataframe
- Não existem valores nulos no Banco de Dados
- Antes da limpeza, 4014 linhas eram duplicadas, agora não existem mais linhas duplicadas.
- Duplicação de linhas X Duplicação de colunas - Quando uma linha é duplicada, temos dados presentes em todas as colunas que são duplicados. A duplicação delas é complicada de lidar, porque ela pode atrapalhar o resultado final da análise, com valores e informações que geram desvios incorretos.

  Quando uma coluna é duplicada, isso significa que ocorre uma redundância do Banco, porque é o mesmo dado em um frequência desnecessária e prejudicial. A duplicação de colunas ocorre quando existe algum erro de código ou caso seja as mesmas informações armazenadas em duas colunas que são também, a mesma coisa.
- Dados duplicados afetam médias, contagens e conclusões, cada valor pode gerar uma alteração que pode ser pequena, mas que gera diferenças do real que podem atrapalhar dependendo do tamanho do banco, sendo uma solução de tamanho empresarial por exemplo.
